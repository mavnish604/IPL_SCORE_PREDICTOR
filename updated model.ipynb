```python
def update_running_score(df):
    df["score"] = 0 
    for (match, innings), group in df.groupby(["matchId", "inning"]):
        total_score = 0 
        for idx in group.index:
            total_score += df.at[idx, "runs_scored"]
            df.at[idx, "score"] = total_score  

    return df
```

```python
df = update_running_score(df)
```

```python
import pandas as pd
import requests
import zipfile
import io
import json

print("Downloading IPL match metadata from Cricsheet...")
url = "https://cricsheet.org/downloads/ipl_json.zip"
headers = {'User-Agent': 'Mozilla/5.0'}

# Download and unzip the files entirely in memory
response = requests.get(url, headers=headers)
z = zipfile.ZipFile(io.BytesIO(response.content))

match_data = []

for filename in z.namelist():
    if filename.endswith(".json") and filename != "README.txt":
        with z.open(filename) as f:
            data = json.load(f)

            match_id = int(filename.split('.')[0])

            venue = data.get('info', {}).get('venue', 'Unknown')
            
            match_data.append({'matchId': match_id, 'venue': venue})

matches_df = pd.DataFrame(match_data)


df = pd.merge(df, matches_df, on='matchId', how='left')

print(f"Missing venues after merge: {df['venue'].isnull().sum()}")
```
